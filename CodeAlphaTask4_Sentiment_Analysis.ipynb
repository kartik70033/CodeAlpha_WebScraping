```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob
```

## 2. Creating the Review Dataset


```python
data = {
    "Review": [
        "This book is absolutely amazing and wonderful!",
        "I really enjoyed reading this book.",
        "The story was fantastic and interesting.",
        "I love this book. It is excellent!",
        "A brilliant and enjoyable story.",
        "This book is terrible and boring.",
        "I hated this book. It was awful.",
        "The story was disappointing and poorly written.",
        "Very bad book. I did not enjoy it.",
        "A horrible and unpleasant reading experience.",
        "The book arrived yesterday.",
        "This book has 250 pages.",
        "The cover of the book is blue.",
        "I received the book today.",
        "The author published this book last year."
    ]
}

df = pd.DataFrame(data)

print("Review Dataset Created Successfully!")
display(df)
```

## 3. Performing Sentiment Analysis

```python
def analyse_sentiment(review):
    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Review"].apply(analyse_sentiment)

print("Sentiment Analysis Completed!")
display(df)
```

## 4. Sentiment Distribution

```python
sentiment_counts = df["Sentiment"].value_counts()

print("Sentiment Analysis Results:")
print(sentiment_counts)
```

## 5. Sentiment Visualization

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Sentiment",
    order=["Positive", "Negative", "Neutral"]
)

plt.title("Sentiment Distribution of Book Reviews")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

plt.show()
```

## 6. Sentiment Percentage Analysis

```python
sentiment_percentage = df["Sentiment"].value_counts(normalize=True) * 100

print("Sentiment Percentage:")
print(sentiment_percentage.round(2))
```

## 7. Sentiment Distribution Pie Chart

```python
plt.figure(figsize=(7, 7))

plt.pie(
    sentiment_percentage,
    labels=sentiment_percentage.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage Distribution of Sentiments")
plt.show()
```

## 8. Review Polarity Analysis

```python
df["Polarity"] = df["Review"].apply(
    lambda review: TextBlob(review).sentiment.polarity
)

print("Review Polarity Scores:")
display(df[["Review", "Polarity", "Sentiment"]])
```

## 9. Review Polarity Visualization

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x=df.index,
    y="Polarity"
)

plt.title("Sentiment Polarity of Individual Reviews")
plt.xlabel("Review Number")
plt.ylabel("Polarity Score")
plt.axhline(y=0, linestyle="--")

plt.show()
```
