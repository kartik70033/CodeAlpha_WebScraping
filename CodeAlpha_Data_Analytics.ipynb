**Objective:** To extract book titles, prices, ratings and availability from a website using Python and organise the collected data into a structured dataset.

## 2. Libraries and Dependencies

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## 3. Website Connection

```python
url = "https://books.toscrape.com/"

response = requests.get(url)

print("Website Status Code:", response.status_code)
```

## 4. HTML Extraction

```python
soup = BeautifulSoup(response.text, "html.parser")

print("Webpage Title:", soup.title.text)
```

## 5. Data Extraction

```python
books = soup.find_all("article", class_="product_pod")

data = []

for book in books:

    title = book.h3.a["title"]

    price = book.find(
        "p", class_="price_color"
    ).text.replace("Â£", "£")

    rating = book.find(
        "p", class_="star-rating"
    )["class"][1]

    availability = book.find(
        "p", class_="instock availability"
    ).text.strip()

    data.append({
        "Title": title,
        "Price": price,
        "Rating": rating,
        "Availability": availability
    })

print("Total books extracted:", len(data))
```

## 6. DataFrame Creation

```python
df = pd.DataFrame(data)

print("Dataset Preview:")

display(df.head())

print("Total Records:", len(df))
```

## 7. Dataset Export


```python
df.to_csv("books_dataset.csv", index=False)

print("Dataset exported successfully!")
```
