## 1. Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Loading the Dataset

```python
from google.colab import files

uploaded = files.upload()
```

```python
df = pd.read_csv("books_dataset.csv")

print("Dataset loaded successfully!")
display(df.head())
```

## 3. Book Rating Distribution

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Rating",
    order=["One", "Two", "Three", "Four", "Five"]
)

plt.title("Distribution of Book Ratings")
plt.xlabel("Book Rating")
plt.ylabel("Number of Books")

plt.show()
```

## 4. Book Price Distribution

```python
# Convert price values into numbers
df["Price"] = df["Price"].str.replace("£", "", regex=False).astype(float)

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Price", bins=8, kde=True)

plt.title("Distribution of Book Prices")
plt.xlabel("Book Price (£)")
plt.ylabel("Number of Books")

plt.show()
```

## 5. Average Book Price by Rating

```python
average_price = df.groupby("Rating")["Price"].mean()

average_price = average_price.reindex(
    ["One", "Two", "Three", "Four", "Five"]
)

plt.figure(figsize=(8, 5))

sns.barplot(x=average_price.index, y=average_price.values)

plt.title("Average Book Price by Rating")
plt.xlabel("Book Rating")
plt.ylabel("Average Price (£)")

plt.show()
```

## 6. Relationship Between Book Price and Rating

```python
df["Rating_Number"] = df["Rating"].map({
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
})

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Rating_Number",
    y="Price"
)

plt.title("Book Price vs Rating")
plt.xlabel("Book Rating")
plt.ylabel("Book Price (£)")

plt.xticks([1, 2, 3, 4, 5])

plt.show()
```

## 7. Book Availability Distribution

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Availability")

plt.title("Book Availability Distribution")
plt.xlabel("Availability Status")
plt.ylabel("Number of Books")

plt.show()
```

## 8. Combined Data Visualization Dashboard

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Rating distribution
sns.countplot(
    data=df,
    x="Rating",
    order=["One", "Two", "Three", "Four", "Five"],
    ax=axes[0, 0]
)
axes[0, 0].set_title("Book Rating Distribution")
axes[0, 0].set_xlabel("Rating")
axes[0, 0].set_ylabel("Number of Books")

# 2. Price distribution
sns.histplot(data=df, x="Price", bins=8, ax=axes[0, 1])
axes[0, 1].set_title("Book Price Distribution")
axes[0, 1].set_xlabel("Price (£)")

# 3. Average price by rating
sns.barplot(
    x=average_price.index,
    y=average_price.values,
    ax=axes[1, 0]
)
axes[1, 0].set_title("Average Price by Rating")
axes[1, 0].set_xlabel("Rating")
axes[1, 0].set_ylabel("Average Price (£)")

# 4. Price vs rating
sns.scatterplot(
    data=df,
    x="Rating_Number",
    y="Price",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Price vs Rating")
axes[1, 1].set_xlabel("Rating")
axes[1, 1].set_ylabel("Price (£)")
axes[1, 1].set_xticks([1, 2, 3, 4, 5])

plt.suptitle("Books Dataset — Data Visualization Dashboard", fontsize=16)
plt.tight_layout()
plt.show()
```
